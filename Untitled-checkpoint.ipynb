# Seaborn

```python
#distplot
#Sales
sns.distplot(Ecommerce.Sales,color='g')
plt.show()

#profit
sns.distplot(Ecommerce.Profit,color='g')
plt.show()
```

```python
#diplot
#sales
sns.displot(Ecommerce.Sales,color='r')
plt.show()

#Profit
sns.displot(Ecommerce.Profit,color='r')
plt.show()
```

```python
#kdeplot
#sales
sns.kdeplot(Ecommerce.Sales,color='b')
plt.show()


#profit
sns.kdeplot(Ecommerce.Profit,color='b')
plt.show()
```

```python
#countplot
#Sales
sns.countplot(Ecommerce.Category,color='g')
plt.show()

#profit
sns.countplot(x=Ecommerce.Profit,Ecommerce.Sales,fill=True)
plt.show()
```

```python
#Box Plot
#Sales
plt.figure(figsize=(8,5))
sns.boxplot(x=Ecommerce.Category,y=Ecommerce.Profit,fill = True)
plt.show()

#profit
sns.boxplot(x=Ecommerce.Profit,y=Ecommerce.Sales,fill = True,hue=Ecommerce.Year)
plt.show()
```
